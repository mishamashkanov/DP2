## 2.2. Проверка предустановленных пакетов

Среда Colab поставляется с предустановленным набором библиотек для анализа данных.
Ниже программно проверяется наличие требуемых пакетов и определяются их версии.

```python
import importlib

REQUIRED = [
    ("numpy",        "numpy"),
    ("pandas",       "pandas"),
    ("matplotlib",   "matplotlib"),
    ("seaborn",      "seaborn"),
    ("scikit-learn", "sklearn"),
]

rows = []
for dist_name, import_name in REQUIRED:
    try:
        module = importlib.import_module(import_name)
        rows.append({"Пакет": dist_name, "Имя при импорте": import_name,
                     "Статус": "предустановлен", "Версия": getattr(module, "__version__", "н/д")})
    except ImportError:
        rows.append({"Пакет": dist_name, "Имя при импорте": import_name,
                     "Статус": "ОТСУТСТВУЕТ", "Версия": "-"})

import pandas as pd

versions_colab = pd.DataFrame(rows)
print("Аудит базовых пакетов анализа данных в среде Google Colab:")
versions_colab
```

## 2.3. Управление аппаратными ускорителями

Colab позволяет выполнять код на трёх типах вычислителей: **CPU** (по умолчанию),
**GPU** (графический ускоритель NVIDIA) и **TPU** (тензорный процессор Google).

Переключение выполняется через меню:

**Среда выполнения → Сменить среду выполнения** (*Runtime → Change runtime type*) →
в поле *Аппаратный ускоритель* (*Hardware accelerator*) выбирается **CPU**, **T4 GPU**
или **TPU** → **Сохранить**.

После смены типа ускорителя виртуальная машина **перезапускается**: все переменные
и установленные через `!pip install` пакеты теряются, поэтому ячейки нужно выполнять
заново.

![Диалог смены среды выполнения](images/05_colab_runtime_type.png)

### Порядок выполнения этого раздела

1. Установить режим **CPU** → выполнить блок «Режим CPU».
2. Переключиться на **T4 GPU** → выполнить блок «Режим GPU».
3. Переключиться на **TPU** → выполнить блок «Режим TPU».

Выводы всех трёх блоков сохраняются в блокноте и служат подтверждением выполнения задания.

### Режим CPU (ускоритель отсутствует)

В этом режиме графический ускоритель не подключён, поэтому утилита `nvidia-smi`
драйвер не обнаруживает — это ожидаемый результат.

```python
print("===== Проверка наличия GPU в режиме CPU =====")
!nvidia-smi || echo "GPU не подключён: аппаратный ускоритель имеет тип CPU"

import multiprocessing
print("\nДоступно логических ядер CPU:", multiprocessing.cpu_count())
```

### Режим GPU

**Среда выполнения → Сменить среду выполнения → T4 GPU → Сохранить.**

Утилита `nvidia-smi` (NVIDIA System Management Interface) выводит модель ускорителя,
версию драйвера и CUDA, объём и текущую занятость видеопамяти, температуру
и энергопотребление.

> **О выводе ячеек этого раздела.** Сведения об оборудовании в режиме GPU были получены при выполнении блокнота с подключённым ускорителем **T4 GPU** — результат приведён скриншотом ниже. При последующем переключении среды выполнения на **TPU** виртуальная машина перезапускается, и вывод ячеек этого раздела замещается сообщением об отсутствии GPU, поэтому он очищен.

```python
# Проверка доступности GPU из PyTorch (предустановлен в Colab)
import torch

print("Версия PyTorch          :", torch.__version__)
print("CUDA доступна           :", torch.cuda.is_available())

if torch.cuda.is_available():
    print("Версия CUDA             :", torch.version.cuda)
    print("Число устройств         :", torch.cuda.device_count())
    print("Модель устройства       :", torch.cuda.get_device_name(0))
    props = torch.cuda.get_device_properties(0)
    print("Объём видеопамяти, ГБ   :", round(props.total_memory / 1024**3, 2))
    print("Мультипроцессоров (SM)  :", props.multi_processor_count)
    print("Compute capability      :", f"{props.major}.{props.minor}")
else:
    print("GPU не подключён — переключите среду выполнения на T4 GPU")
```

```python
# Практическая проверка ускорения: умножение больших матриц на CPU и на GPU
import torch, time

size = 4000
a_cpu, b_cpu = torch.randn(size, size), torch.randn(size, size)

t0 = time.perf_counter()
_ = a_cpu @ b_cpu
cpu_time = time.perf_counter() - t0
print(f"Умножение матриц {size}x{size} на CPU : {cpu_time:.3f} с")

if torch.cuda.is_available():
    a_gpu, b_gpu = a_cpu.cuda(), b_cpu.cuda()
    torch.cuda.synchronize()
    t0 = time.perf_counter()
    _ = a_gpu @ b_gpu
    torch.cuda.synchronize()
    gpu_time = time.perf_counter() - t0
    print(f"Умножение матриц {size}x{size} на GPU : {gpu_time:.3f} с")
    print(f"Ускорение: в {cpu_time / gpu_time:.1f} раза")
else:
    print("GPU недоступен — сравнение не выполнено")
```

### Режим TPU

**Среда выполнения → Сменить среду выполнения → TPU → Сохранить.**

TPU (Tensor Processing Unit) — специализированный процессор Google для тензорных
вычислений. В отличие от GPU, он не виден утилите `nvidia-smi`: устройства TPU
отображаются в системе как `/dev/accel*`, а работа с ними ведётся через библиотеки
`torch_xla` или `jax`.

```python
import os

print("===== Признаки подключённого TPU =====")
print("Переменная COLAB_TPU_ADDR :", os.environ.get("COLAB_TPU_ADDR", "не задана"))
print("Переменная TPU_NAME       :", os.environ.get("TPU_NAME", "не задана"))
print("\nУстройства-ускорители в системе:")
!ls -l /dev/accel* 2>/dev/null || echo "устройства /dev/accel* не найдены"
```

```python
# Информация об устройствах TPU через JAX (предустановлен в Colab)
try:
    import jax
    print("Версия JAX        :", jax.__version__)
    devices = jax.devices()
    print("Тип платформы     :", devices[0].platform.upper())
    print("Число устройств   :", len(devices))
    print()
    for i, d in enumerate(devices):
        print(f"  Устройство {i}: {d}")
        print(f"     вид (kind)        : {getattr(d, 'device_kind', 'н/д')}")
        print(f"     идентификатор     : {getattr(d, 'id', 'н/д')}")
except Exception as e:
    print("Не удалось получить сведения через JAX:", type(e).__name__, e)
```

```python
# Альтернативный способ — через PyTorch/XLA
try:
    import torch_xla
    import torch_xla.core.xla_model as xm

    print("Версия torch_xla        :", torch_xla.__version__)
    device = xm.xla_device()
    print("Устройство XLA          :", device)
    print("Тип устройства          :", xm.xla_device_hw(device))
    print("Число ядер TPU          :", xm.xrt_world_size() if hasattr(xm, "xrt_world_size") else "н/д")
except Exception as e:
    print("torch_xla недоступен либо TPU не подключён:", type(e).__name__, e)
    print("Убедитесь, что в меню «Сменить среду выполнения» выбран ускоритель TPU")
```
